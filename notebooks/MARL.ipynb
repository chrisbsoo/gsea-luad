```python
import pandas as pd

df = pd.read_csv("../data/gsea_cleaned.csv")
df.head()
```

```python
from scipy import stats

import pandas as pd
from scipy import stats

def differential_expression_anova(df, group_col):
    results = []
    
    groups = df[group_col].unique()
    
    genes = df.drop(columns=[group_col]).columns
    
    for gene in genes:
        group_data = [
            df[df[group_col] == g][gene].dropna() 
            for g in groups
        ]
        
        f_stat, p_value = stats.f_oneway(*group_data)
        
        results.append({
            'gene': gene,
            'f_statistic': f_stat,
            'p_value': p_value
        })
        
    results_df = pd.DataFrame(results).set_index('gene')
    return results_df

df_t = df.drop("Samples", axis=1)
result = differential_expression_anova(df_t, "C_STAGE")
```

```python
sig_genes = result[result["p_value"] < 0.05]
print(sig_genes)
```
